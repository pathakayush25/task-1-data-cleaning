import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("titanic_raw.csv")

# Display basic information
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Missing value percentage
missing = df.isnull().sum()
missing_percentage = (missing / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percentage
})

print("\nMissing Value Summary:")
print(missing_table)

# Check duplicates
print("\nDuplicate Records:", df.duplicated().sum())

# Check data types
print("\nData Types:")
print(df.dtypes)

# Remove duplicate records
df.drop_duplicates(inplace=True)

# Convert numeric columns
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Fare'] = pd.to_numeric(df['Fare'], errors='coerce')

# Handle missing values
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['Cabin'] = df['Cabin'].fillna('Unknown')

# Clean categorical/text values
df['Name'] = df['Name'].str.strip()
df['Sex'] = df['Sex'].str.strip().str.lower()
df['Ticket'] = df['Ticket'].str.strip()
df['Embarked'] = df['Embarked'].str.strip().str.upper()

# Check invalid values
print("\nInvalid Age Records:")
print(df[df['Age'] < 0])

print("\nInvalid Fare Records:")
print(df[df['Fare'] < 0])

print("\nInvalid Survived Records:")
print(df[~df['Survived'].isin([0, 1])])

print("\nInvalid Pclass Records:")
print(df[~df['Pclass'].isin([1, 2, 3])])

# Final checks
print("\nFinal Missing Values:")
print(df.isnull().sum())

print("\nFinal Duplicate Count:")
print(df.duplicated().sum())

print("\nFinal Data Types:")
print(df.dtypes)

# Save cleaned dataset
df.to_csv("titanic_cleaned.csv", index=False)

print("\nCleaned dataset saved successfully.")
